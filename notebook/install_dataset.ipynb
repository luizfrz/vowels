```python
import kagglehub
import os
import shutil
```

- Uma variavel definindo onde está pasta data
- Faz uma validação, caso não existir nenhuma pasta cria uma

```python
data_dir = '../data'
cache_path  = kagglehub.dataset_download("crawford/emnist")
if not os.path.exists(data_dir):
    os.makedirs(data_dir, exist_ok=True)
    print("Pasta criada com sucesso.", os.listdir(data_dir) )
else:
    print("Pasta existe:", os.listdir(data_dir)) 
    exit
```

- Lê a pasta de cache completa listando todos os nomes de arquivos e subpastas para a memória.
- Constrói o caminho completo de origem juntando o diretório de cache ao nome do item atual.
- Verifica se o item de origem atual se trata de um diretório (pasta) completo ou não.
- Copia a pasta inteira com todas as subpastas e arquivos internos, ignorando erros se o destino já existir.
- Se o item atual não for uma pasta (ou seja, se for apenas um arquivo individual).
- Faz a cópia direta do arquivo isolado da origem para a pasta de destino final.


```python
for filename in os.listdir(cache_path):
    origin = os.path.join(cache_path, filename)
    destin = os.path.join(data_dir, filename)
    
    if os.path.isdir(origin):
        shutil.copytree(origin, destin, dirs_exist_ok=True)
    else:
        shutil.copy(origin, destin)
```

- Imprime onde o caminho que dataset foi salvo

```python
print("Dataset baixado com sucesso", os.path.abspath(data_dir))
```
