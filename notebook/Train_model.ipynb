```python
import os
import shutil
import json
from tqdm import tqdm
from sklearn.model_selection import train_test_split
```

```python
DATA_DIR = "/kaggle/input/datasets/mohamedaoul/synthetic-handwritten-uppercase-a-z-28x28/dataset_handwritten_28x28"

OUTPUT = "/kaggle/working/dataset"

TRAIN_DIR = os.path.join(OUTPUT, "train")
TEST_DIR = os.path.join(OUTPUT, "test")

os.makedirs(TRAIN_DIR, exist_ok=True)
os.makedirs(TEST_DIR, exist_ok=True)
```

```python

for char in classes:

    src_folder = os.path.join(
        SOURCE,
        char
    )

    images = [
        x for x in os.listdir(src_folder)
        if x.lower().endswith((".png", ".jpg", ".jpeg"))
    ]

    train_imgs, test_imgs = train_test_split(
        images,
        test_size=0.2,
        random_state=42
    )


    train_class = os.path.join(
        TRAIN_DIR,
        char
    )

    test_class = os.path.join(
        TEST_DIR,
        char
    )

    os.makedirs(train_class, exist_ok=True)
    os.makedirs(test_class, exist_ok=True)


    for img in tqdm(train_imgs, desc=f"Train {char}"):

        shutil.copy(
            os.path.join(src_folder, img),
            os.path.join(train_class, img)
        )


    for img in test_imgs:

        shutil.copy(
            os.path.join(src_folder, img),
            os.path.join(test_class, img)
        )
```
